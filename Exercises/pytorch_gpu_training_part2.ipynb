# Exercise Part 2: Requesting GPU Resources and Running PyTorch
In this part, we will demonstrate how to request GPU resources on the Explorer Cluster using SLURM and run your PyTorch training script or notebook on the GPU.

## Step 1: Loading Required Modules and Setting Up the Environment
Before requesting GPU resources, ensure that the required modules are loaded, and the environment is set up properly.
Run the following commands in the terminal or notebook to load the required modules and create a new Conda environment.

```python
# Loading the required modules
%%bash
module load python/3.10   # or: module load anaconda3   (see conda_environment_setup_part1.ipynb for venv vs conda)

# Recommended: venv in scratch space for the current user (fast, portable)
python -m venv /scratch/$USER/pytorch_env_course
source /scratch/$USER/pytorch_env_course/bin/activate

# Install JupyterLab
pip install jupyterlab

# Install PyTorch with CUDA 12.x support (matches Explorer's H200/A100 driver, CUDA 12.8)
pip install torch torchvision torchaudio --index-url https://download.pytorch.org/whl/cu121
```

## Step 2: Requesting GPU Resources
Use SLURM to request GPU resources for training. The following commands are used to request GPU nodes during and after training.


```python
##### Use after training ######
%%bash
srun --partition=gpu --nodes=1 --cpus-per-task=1 --gres=gpu:1 --mem=2G --time=00:05:00 --pty /bin/bash
################################

##### Use during training (short interactive dev/test) ######
srun --partition=gpu --nodes=1 --cpus-per-task=4 --gres=gpu:1 --mem=16G --time=00:30:00 --pty /bin/bash
################################

# Requesting a GPU node
srun --partition=gpu --nodes=1 --cpus-per-task=1 --gres=gpu:1 --mem=2G --time=00:05:00 --pty /bin/bash
nvidia-smi
exit

# Requesting a specific GPU type -- Explorer's flagship is the H200 (141GB)
srun --partition=gpu --nodes=1 --cpus-per-task=1 --gres=gpu:h200:1 --mem=2G --time=00:05:00 --pty /bin/bash
nvidia-smi
exit

# Check available features on the GPU partition
sinfo -p gpu --Format=nodes,cpus,nodelist,gres,features
sinfo -p gpu --Format=nodes,cpus,nodelist,gres,statecompact,features
```

These commands will allow you to request GPU resources for running your PyTorch models on the Explorer Cluster.
